```python
from pathlib import Path
import pandas as pd

# Configuration
SAMPLE_SIZE = 100
RANDOM_SEED = 42

# GitHub repository
GITHUB_BASE = (
    "https://raw.githubusercontent.com/semoglou/cart/"
    "main/llm_baselines/evaluation/"
)

# Output directory in Colab
SAMPLE_DIR = Path("/content/samples") / f"sample_{SAMPLE_SIZE}"
SAMPLE_DIR.mkdir(parents=True, exist_ok=True)

# Load full evaluation files directly from GitHub
evaluation_input = pd.read_csv(
    GITHUB_BASE + "evaluation_input.csv"
)

evaluation_mapping = pd.read_csv(
    GITHUB_BASE + "evaluation_mapping.csv"
)

# Validate IDs
assert evaluation_input["annotation_id"].is_unique
assert evaluation_mapping["annotation_id"].is_unique
assert SAMPLE_SIZE <= len(evaluation_input)

# Reproducible random sample
sampled_input = (
    evaluation_input
    .sample(n=SAMPLE_SIZE, random_state=RANDOM_SEED)
    .reset_index(drop=True)
)

# Corresponding mapping, preserving sample order
sampled_mapping = (
    evaluation_mapping
    .set_index("annotation_id")
    .loc[sampled_input["annotation_id"]]
    .reset_index()
)

# Validate correspondence
assert (
    sampled_input["annotation_id"].tolist()
    == sampled_mapping["annotation_id"].tolist()
)

# Save sampled CSV files
input_path = SAMPLE_DIR / f"evaluation_input_sample{SAMPLE_SIZE}.csv"
mapping_path = SAMPLE_DIR / f"evaluation_mapping_sample{SAMPLE_SIZE}.csv"

sampled_input.to_csv(input_path, index=False, encoding="utf-8-sig")
sampled_mapping.to_csv(mapping_path, index=False, encoding="utf-8-sig")

print(f"Sample size: {len(sampled_input)}")
print(f"Random seed: {RANDOM_SEED}")
print(f"Unique IDs: {sampled_input['annotation_id'].nunique()}")
print(f"\nSaved input:   {input_path}")
print(f"Saved mapping: {mapping_path}")

display(sampled_input.head())
```
